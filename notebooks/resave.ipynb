```python
import pandas as pd


SEED = 42


def print_memory_usage(df: pd.DataFrame) -> None:
    print(f"{df.memory_usage(deep=True).sum() / 1024**2} mb\n")
    return None
```

```python
rating = pd.read_csv(
    "../data/raw/ml-25m/ratings.csv",
)
print(rating.dtypes)
print_memory_usage(rating)
```

```python
rating = rating.astype(
    {
        'movieId': 'int32',
        'rating': 'float32',
        'userId': 'int32'
    }
)
rating['timestamp'] = pd.to_datetime(
    rating['timestamp'],
    unit='s'
)
```

```python
print(rating.dtypes)
print_memory_usage(rating)
```

```python
rating.to_parquet("../data/processed/rating.parquet")

del rating
```
